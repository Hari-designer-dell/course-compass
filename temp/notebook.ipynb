```python
def content_rec_df(course, top_n=10):
    #clean data
    processed_course = nfx.remove_stopwords(course)
    processed_course = nfx.remove_special_characters(processed_course)

    #check if course still exists in dataframe
    if processed_course not in data['course_title'].values:
        print(f"Course '{course}' not found.")
        return pd.DataFrame()

    course_index = data[data['course_title'] == processed_course].index[0]
    distances = similarity[course_index]
    top_courses = sorted(enumerate(distances), reverse=True, key=lambda x: x[1])[1:top_n+1]

    recs_df = pd.DataFrame({
        'course_title': [data.iloc[i[0]].course_title for i in top_courses],
        'similarity_score': [i[1] for i in top_courses],
        'num_subscribers': [data.iloc[i[0]].num_subscribers for i in top_courses],
        'num_reviews': [data.iloc[i[0]].num_reviews for i in top_courses],
        'pop_score': [data.iloc[i[0]].pop_score for i in top_courses]
    })

    return recs_df
```

```python
recs_df = content_rec_df('Ultimate Investment Banking Course')
recs_df
```

```python
numerical_features = recs_df[['similarity_score', 'num_subscribers', 'num_reviews', 'pop_score']].values
```

```python
cv = CountVectorizer(max_features=3000)
text_vectors = cv.fit_transform(recs_df['course_title'])
```

```python
X = hstack([text_vectors, numerical_features])
```

```python
def create_labels(row):
    if row['similarity_score']>=0.9:
        return 2
    if row['similarity_score']>=0.3:
        return 1
    else:
        return 0
```

```python
recs_df['relevance'] = recs_df.apply(create_labels, axis=1)
y = recs_df['relevance']
```

```python
recs_df
```

```python
rf = RandomForestClassifier(random_state=42)
rf.fit(X, y)
```

```python
predictions = rf.predict(X)
recs_df['predicted_suitability'] = predictions
```
